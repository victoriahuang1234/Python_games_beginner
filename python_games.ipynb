Rolling Dice Game

```python
#count how many time user put "y"
#Loop infinite
# Ask:roll the dice?
#if user enters y
#   ask how many die to roll
#   Loop until u roll input number of die
#       generate a random number 
#       add the random number to list of random number    
#  print list
#If user enters n
#  print thank you message and how many times user has rolled
#  terminate
#else
#print invalid choice
```

```python

import random
#starting the loop 
counter=0
while True:
#we have to convert whatever the user input as lowercase
    user_choice = input("Roll the dice? (y/n): ").lower()
    if user_choice== "y":
        counter +=1
        numdie=int(input("How many dice do you want to roll?"))
        results=[]
        while len(results) < numdie :
    #this returns a random number between 1 & 6
            die1= random.randint(1,6)
            results.append(die1)
        print(f" {results}")
    elif user_choice=="n":
        print(f"Thanks for playing! You rolled {counter} times this session." )
        break
    else:
        print('Invalid Choice')

```

```python
#import random
#assign a variable a random number 
# if user iputs not int (use try & except block)
#  return please enter a valid number
#if user ipunts<var:
#  you guessed too low
#if user iputs>var:
#  you guessed too high 
#if user inputs==var:
#   print u have guessed the right number
#   terminate
```

```python
import random
num= random.randint(0,100)
counter=0

while True:
    #try and except blocks are used to handle runtime errors (exceptions) so a program 
    # can recover gracefully instead of crashing abruptly. You place the risky code 
    # inside the try block, and the fallback response inside the except block.
    try:
        guess=int(input('Guess the number between 0 to 100'))   
    #do not use if not guess because they evaluate 0 as a False instead of a number
    except ValueError:
        print ('Invalid input, please enter a number between 0 to 100')
        continue
    counter +=1    
    if guess>100 or guess<0:
        print ("guess between 0 to 100")
    elif guess > num:
        print("Guess lower")
    elif guess < num:
        print("Guess Higer")
    else:
        print (f"Hooray you got it! Its {num}! You took {counter}tries")
        break
        
    
```

Rock Paper or Scissors

```python
#Ask user to make a choice
#If choice is not valid
#   Print an error
#Let computer make a choice
#Determine the winner
#Ask user if they want to continue 
#if not
#   terminat
```

```python
import random

user_choice = input("Choose r/p/s: ").lower()
# Get the computer choice using correct syntax
comp_choice = random.choice(["r", "p", "s"])

# Check if the input is valid using 'in'
if user_choice in ["r", "p", "s"]:
    
    print(f"Computer chose: {comp_choice}")
    
    if comp_choice == user_choice:
        print("Draw!")
    elif comp_choice == "r" and user_choice == "p":
        print("Win!")
    elif comp_choice == "p" and user_choice == "s":
        print("Win!")
    elif comp_choice == "s" and user_choice == "r":
        print("Win!")
    elif comp_choice == "r" and user_choice == "s":
        print("Lose!")
    elif comp_choice == "p" and user_choice == "r":
        print("Lose!")
    elif comp_choice == "s" and user_choice == "p":
        print("Lose!")
else:
    print("Invalid Choice!")
    
```

Modularising Code: breaking down a code to smaller resuable functions
Modularise from the RPS game

```python
import random
ROCK="r"
PAPER="p"
SCISSORS="s"
def get_user_choice():
    # Use a loop to keep asking until the user provides a valid input
    while True:
        choice = input("Choose r/p/s: ").lower()
        if choice in [ROCK, PAPER, SCISSORS]:
            return choice
        print("Invalid Choice! Please try again.")

def get_computer_choice():
    # Return the computer's random selection
    return random.choice([ROCK, PAPER, SCISSORS])

def determine_winner(user_choice, comp_choice):
    # Consolidate the logic to compare choices and return the outcome
    print(f"Computer chose: {comp_choice}")
    
    if user_choice == comp_choice:
        return "Draw!"
    # Grouping the winning conditions together makes the code much cleaner
    elif (user_choice == PAPER and comp_choice == ROCK) or \
         (user_choice == SCISSORS and comp_choice == PAPER) or \
         (user_choice == ROCK and comp_choice == SCISSORS):
        return "Win!"
    else:
        # If it's not a draw and you didn't win, you logically must have lost
        return "Lose!"

def play_game():
    # This is the main controller that calls the other modules in order
    user = get_user_choice()
    comp = get_computer_choice()
    result = determine_winner(user, comp)
    print(result)

# Execute the program by calling the main function
play_game()
```
